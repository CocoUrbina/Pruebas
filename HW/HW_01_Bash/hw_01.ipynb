Con el siguiente codigo:

```bash
#!/bin/bash

residuo=$((${2}%10))
parteEntera=$((${2}/10))

#echo $residuo
#echo $parteEntera
  
for ((j=1;j<=residuo;j++));do #agrega numeros de 1 en 1
  ./random.sh | head -n 4 | tail -n 1 >> rand_${1}_${2}.dat
done
#ciclo desde 1 hasta residuo, inclusives. Va de 1 en 1.
#./random.sh: ejecuta el script random.sh
#head -n 4: toma solamente las primeras 4 filas de outcome (1.linea en blanco, 2."10 ran...", 3. ---, 4. <primer numero>
#tail -n 1: toma solamente la ultima fila del outcome anterior. Es decir <primer numero>.
#>> rand_${1}_${2}.dat: el outcome se agrega a archivo rand_${1}_${2}.dat

for ((i=1; i<=parteEntera;i++));do #agrega numeros de 10 en 10
  ./random.sh | head -n 13 | tail -n 10  >> rand_${1}_${2}.dat
done
#ciclo desde 1 hasta parteEntera, inclusives. Va de 1 en 1.
#./random.sh: ejecuta el script random.sh
#head -n 13: toma las primeras 13 filas de outcome (1.linea en blanco, 2."10 ran...", 3. ---, 4. <primer numero> ... 13.<decimo numero>
#tail -n 10: toma las ultimas 10 filas del outcome anterior. Es decir 4. <primer numero> ... 13.<decimo numero>.
#>> rand_${1}_${2}.dat: el outcome se agrega a archivo rand_${1}_${2}.dat
```

### 2. Se le ha entregado un archivo llamado `usuarios.csv` que contiene una lista de 25k nombres, apellidos, correos electrónicos y lugar de trabajo. Para ninguna de las siguientes tareas es permitido usar for loops.

#### 2.a) Genere una lista de todos los usuarios que se llaman *Brian*, cuyo correo electrónico termina en *.com* y que cooperan con la compañı́a *Boeing*. Explique su razonamiento. **(0.5%)**

Las filas del archivo tienen la forma:
Nombre,Apellido,email,compania. Entonces la busqueda se puede hacer con un pipeline de dos `grep`.

`grep '^Brian' usuarios.csv | grep '.*.com,Boeing'`

Donde:

`grep '^Brian' usuarios.csv` filtra todas las filas que inician con **Brian**

`grep '.*.com,Boeing'` filtraa todas las filas que terminan con **.com,Boeing**

#### 2.b) Genere una lista de todos los usuarios que se llaman *James* y *Paul*, pero solo liste aquellos usuarios cuyo apellido empieza con *J* o *S* y cooperan con *Ad Astra* **(0.5%)**

```bash
grep -E '^James,J|^James,S|^Paul,J|^Paul,S' usuarios.csv | grep '.*Astra'

```

Donde

`-E` es para extender las expresiones, `^` indica inicio y el `|` funciona como un OR.

Luego, `grep '.*Astra'` filtra los que terminan con **Astra**.

#### 2.d) Añada un nuevo campo (field) a la base de datos. A cada entrada, agregue el número de teléfono. Para esta tarea puede añadir a todos los usuarios el número telefónico 5555-5555. Note que la primera lı́nea contiene la descripción de cada campo. No incluya esta lı́nea, solo las demás. En la primera lı́nea se debe añadir el campo *“phone_number”*. Explique su razonamiento. **(0.5%)**

```bash
awk '{print "5555-5555"}' usuarios.csv > columna_telefonos.txt
```

Imprime "5555-5555" en cada fila, tantas veces como la cantidad de filas de `usuarios.csv.`
Luego el resultado lo guarda en un archivo `columna_telefonos.txt`.

Para modificar la primera fila de `columna_telefonos.txt`, se hace como sigue

```bash
sed -i '1s/.*/phone_number/' columna_telefonos.txt
```

Donde la estructura de `sed` es 's/reemplazado/reemplazante/'.

La `s` es por **substitute**.

Para nuestro caso,

el `1` indica la primera fila

el `.*` se puede leer como todo lo que esta antes de nada en esa fila. O sea, todo. Todo en esa fila sera reemplazado.

y el `phone_number` es el reemplazante.

Finalmente, se concatenan el archivo original y la nueva columna y se guardan en un nuevo archivo.

```bash
paste -d',' usuarios.csv columna_telefonos.txt > usuarios_telefonos.csv
```
