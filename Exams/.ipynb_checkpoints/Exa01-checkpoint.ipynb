
¿Qué hace la operación `*pBeta = 0.1;`? ¿`pBeta` se almacena en el stack o en el heap?

3. ¿En cuál situación falla fundamentalmente el método de Newton para encontrar raíces de ecuaciones no lineales?

4. ¿Cuál es la diferencia entre un contructor de copia y la sobrecarga del operador de asignación?

5. Verdadero o falso: para disminuir el error total de las diferencias finitas basta con disminuir $h$, el parámetro de control de la aproximación, de forma arbitraria. **Explique**. Recuerde que $h$ es el "paso" de la diferencia finita.

6. ¿En cuál situación se usa un ajuste polinomial y en cuál se usa una interpolación?

7. En Python, ¿a qué se refiere mutabilidad? Dé un ejemplo de una estructura primitiva mutable y una inmutable en Python.

8. Dé un ejemplo y explique una situación en la cual debemos utilizar el heap para almacenar alguna estructura de datos.

# Segunda Parte: Desarrollo (60%)

## 1. (20%) Derive el error de aproximación y el error de redondeo del método de diferencias finitas hacia adelante. Si consideramos la suma de estos dos errores como el error total, ¿qué conclusión obtiene con respecto al error total de las diferencias finitias? Recuerde que $$\frac{d f(x)}{dx} \approx \frac{f(x+h) - f(x)}{h} \tag{2.1}$$ donde $h$ es el parámetro de control de la aproximación.

Haciendo la expansion de Taylor para $f(x+h)$,

\begin{equation*}
    f(x+h) =  f(x) +hf^\prime(x) + \frac{h^2}{2}f^{\prime \prime}(x) + \mathcal{O}(h^3) \\
\end{equation*}

Despejando $f^\prime(x)$,

\begin{equation*}
    f^\prime(x) =  \frac{f(x+h) - f(x) }{h}  -
    \underbrace{\frac{h}{2}f^{\prime \prime}(x) + \cdots}_{\text{error de aproximacion}}
        \\
\end{equation*}

Omitiendo lo terminos de mayor orden, el error de aproximacion escala linealmente con $h$.

Ahora, en Python se aproxima $f(x)$ con una precision de redondeo de $Cf(x)$, donde $C \sim 10^{-16}$. Por lo que 

\begin{equation*}
   \frac{f(x+h) - f(x) }{h} \sim \frac{2C |f(x)|}{h} 
\end{equation*}

Dicho esto, el error total (error de aproximacion + error de redondeo) viene dado por

\begin{equation*}
    \epsilon(h) = \frac{2C|f(x)|}{h} + \frac{1}{2}h|f''(x)|
\end{equation*}

Derivando con respecto a $h$ e igualando a cero,
 
$$
-\frac{2C|f(x)|}{h^2} + \frac{1}{2}|f''(x)| = 0.
$$

Lo que lleva a 

$$
h_{\rm min} = \sqrt{4C\frac{|f(x)|}{|f''(x)|}},
$$

Luego, haciendo $\epsilon (h_{\rm min})$,

$$
\epsilon_{\rm{min}} = \sqrt{4C|f(x)||f''(x)|}.
$$

Para $|f(x)| \sim |f''(x)| \sim 1$,

$$
\epsilon_{\rm{min}} \sim \sqrt{4C} \sim 2 \sqrt{10^{-16}} \sim \times 10^{-8}.
$$


Se concluye entonces que para diferenias finitas, el error total minimo es de $\sim 10^{-8}$.

## 2. (20%) Utilizando un procedimiento como el que utilizamos en clase, encuentre una aproximación para $$f^{\prime \prime} (x) = \frac{d^2 f(x)}{dx^2} \tag{2.2}$$ 
## Es decir, encuentre una expresión como la mostrada en la ecuación (2.1), pero para la segunda derivada. Note que para la primera derivada, la aproximación depende solamente de la función $f$ evaluada en dos puntos. Para el caso de la segunda derivada, ¿se cumple que la aproximación depende de la función evaluada en dos puntos, o se necesitan más? ¿De qué orden con respecto a $h$ es el error de aproximación en este caso?

Haciendo las expansiones de Taylor para $f(x+h)$ y $f(x-h)$,

\begin{align*}
    f(x+h) &=  f(x) +hf^\prime(x) + \frac{h^2}{2}f^{\prime \prime}(x) + \frac{h^2}{6}f^{\prime \prime \prime}(x) + \mathcal{O}(h^4) \\
    f(x-h) &=  f(x) -hf^\prime(x) + \frac{h^2}{2}f^{\prime \prime} (x) - \frac{h^2}{6}f^{\prime \prime \prime}(x) + \mathcal{O}(h^4) 
\end{align*}

Sumando ambas expresiones,

\begin{equation*}
    f(x+h) + f(x-h) =  2f(x) + h^2f^{\prime \prime}(x) + 2\mathcal{O}(h^4)
\end{equation*}

Despejando $f^{\prime \prime}(x)$,

\begin{equation*}
    f^{\prime \prime}(x) = \frac{f(x-h) -2f(x) + f(x+h)}{h^2} + 2\mathcal{O}(h^2)
\end{equation*}

Se requieren 3 puntos y el error de aproximacion escala con $h^2$.

**Nota**: Era de intuir que los puntos requeridos son 3, pues $f^{\prime \prime}(x)$ es una concavidad.

## 3. (20%) Al realizar la aproximación de una integral definida usando la regla de Simpson, obtenemos que
   
$$(a,b) = \int_a^b dx\ f(x) \approx \frac{h}3\left[f(a) + f(b) + 4\sum_{\substack{k\ impar\\ 1\dots{}N-1}}f(a+kh) + 2\sum_{\substack{k\ par \\ 2\dots{}N-2}}f(a+kh)\right] \tag{2.3}$$

## Escriba una función en `C++` que realiza la aproximación. No debe escribir un `main()` ni incluir bibliotecas, solamente escribir la función que realiza la aproximación de Simpson.

```C++
double Simpson(double(*func)(double), double a, double b, int N) {

    double h = (b - a) / N;
    double sumImpares = 0.0;
    double sumPares = 0.0;

    for(int k = 1; k < N; k = k + 2) {
        sumPares += func(a + k * h);
    }

    for(int k = 2; k < N; k = k + 2) {
        sumImpares += func(a + k * h);
    }
    
    return h * (func(a) + func(b) + 4 * sumImpares + 2 * sumPares) / 3;
```
