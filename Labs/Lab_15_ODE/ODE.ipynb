# Ecuaciones Diferenciales Ordinarias (ODEs)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#ODE

def ode(t, x):
    return -x**3 +np.sin(t)
```

```python
# IVP

t_0 = 0
x_0 = 0
```

```python
# times for N = 20

timesN20 = np.linspace(0.0, 10.0, num = 21)
print(timesN20)
```

```python
# mesh for N = 20

h20 = timesN20[1] - timesN20[0]
print(h20)
```

```python
# times for N = 1000

timesN1000 = np.linspace(0.0, 10.0, num = 1001)
print(timesN1000)
```

```python
# mesh for N = 1000

h1000 = timesN1000[1] - timesN1000[0]
print(h1000)
```

## Euler

```python
def euler(func, t, x, h):
    return x + h * func( t, x)
```

```python
xVals20_euler = np.zeros(timesN20.size) # an array that will store all the values for x, for N = 20
xVals20_euler[0] = 0 # the inicial value

for t in range(timesN20.size - 1): # range va de 0 hasta arg -1. En este caso, desde 0 hasta (21- 1) - 1.
    xVals20_euler[t + 1] = euler(ode, timesN20[t], xVals20_euler[t], h20) # xVals20 va desde _1 hasta _20.
```

```python
xVals1000_euler = np.zeros(timesN1000.size)
xVals1000_euler[0] = 0
for t in range(timesN1000.size - 1):
    xVals1000_euler[t + 1] = euler(ode, timesN1000[t], xVals1000_euler[t], h1000)
```

```python
plt.figure(figsize=(6, 4))
plt.plot(timesN20, xVals20_euler,label="N = 20")
plt.plot(timesN1000, xVals1000_euler,label="(N = 1000")


plt.xlabel("t")
plt.ylabel("x(t)")
plt.title("Aproximación de la solución de $dx/dt = -x^3 + sin(t)$ con Euler")
plt.legend()
plt.grid()
plt.show()
```

## RK2

```python
def rk2(func, t, x, h):
    k_1 = h * func(t, x);
    k_2 = h * func(t + h/2.0, x + k_1/2.0)
    return x + k_2
```

```python
xVals20_rk2 = np.zeros(timesN20.size) # an array that will store all the values for x, for N = 20
xVals20_rk2[0] = 0 # the inicial value

for t in range(timesN20.size - 1): # range va de 0 hasta arg -1. En este caso, desde 0 hasta (21- 1) - 1.
    xVals20_rk2[t + 1] = rk2(ode, timesN20[t], xVals20_rk2[t], h20) # xVals20 va desde _1 hasta _20.
```

```python
xVals1000_rk2 = np.zeros(timesN1000.size)
xVals1000_rk2[0] = 0
for t in range(timesN1000.size - 1):
    xVals1000_rk2[t + 1] = rk2(ode, timesN1000[t], xVals1000_rk2[t], h1000)
```

```python
plt.figure(figsize=(6, 4))
plt.plot(timesN20, xVals20_rk2, label="N = 20")
plt.plot(timesN1000, xVals1000_rk2, label="N = 1000")


plt.xlabel("t")
plt.ylabel("x(t)")
plt.title("Aproximación de la solución de $dx/dt = -x^3 + sin(t)$ con RK2")
plt.legend()
plt.grid()
plt.show()
```

## RK4

```python
def rk4(func, t, x, h):
    k_1 = h * func(t, x);
    k_2 = h * func(t + h/2.0, x + k_1/2.0)
    k_3 = h * func(t + h/2.0, x + k_2/2.0)
    k_4 = h * func(t + h, x + k_3)
    return x + (1/6.0)*(k_1 + 2*k_2 + 2*k_3 + k_4)
```

```python
xVals20_rk4 = np.zeros(timesN20.size) # an array that will store all the values for x, for N = 20
xVals20_rk4[0] = 0 # the inicial value

for t in range(timesN20.size - 1): # range va de 0 hasta arg -1. En este caso, desde 0 hasta (21- 1) - 1.
    xVals20_rk4[t + 1] = rk4(ode, timesN20[t], xVals20_rk4[t], h20) # xVals20 va desde _1 hasta _20.
```

```python
xVals1000_rk4 = np.zeros(timesN1000.size)
xVals1000_rk4[0] = 0
for t in range(timesN1000.size - 1):
    xVals1000_rk4[t + 1] = rk4(ode, timesN1000[t], xVals1000_rk4[t], h1000)
```

```python
plt.figure(figsize=(6, 4))
plt.plot(timesN20, xVals20_rk4, label="N = 20")
plt.plot(timesN1000, xVals1000_rk4, label="N = 1000")


plt.xlabel("t")
plt.ylabel("x(t)")
plt.title("Aproximación de la solución de $dx/dt = -x^3 + sin(t)$ con RK4")
plt.legend()
plt.grid()
plt.show()
```
